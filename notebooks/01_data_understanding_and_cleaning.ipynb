## Data Cleaning

After inspecting the dataset, a separate copy is created for cleaning. The original dataset is kept unchanged so that the cleaning process can be verified.

Missing values, category formatting, and invalid values are handled based on the observations from the data-quality checks.

```python
# Create a separate copy for cleaning
df_clean = df.copy()

# Store original row and column counts
original_rows = len(df)
original_columns = len(df.columns)

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape before cleaning:", df_clean.shape)
```

```python
# Identify text columns
text_columns = df_clean.select_dtypes(include=['object']).columns

# Remove extra spaces from text values
for col in text_columns:
    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
```

```python
# Numeric columns with missing values
numeric_missing_columns = [
    'Delivery_person_Age',
    'Delivery_person_Ratings',
    'multiple_deliveries'
]

# Fill missing numeric values using median
for col in numeric_missing_columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())


# Text columns with missing values
categorical_missing_columns = [
    'Time_Orderd',
    'Weather_conditions',
    'Road_traffic_density',
    'Festival',
    'City'
]

# Fill missing text values with Unknown
for col in categorical_missing_columns:
    df_clean[col] = df_clean[col].fillna('Unknown')


print("Total missing values after cleaning:", df_clean.isnull().sum().sum())
```

```python
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("\nMissing values in cleaned dataset:")
print(df_clean.isnull().sum())

print("\nDuplicate rows in cleaned dataset:", df_clean.duplicated().sum())
print("Duplicate IDs in cleaned dataset:", df_clean['ID'].duplicated().sum())
```

### Cleaning Verification

The cleaned dataset was compared with the original dataset. The original dataset was preserved, and cleaning was performed on a separate copy.

Missing values were handled using median imputation for numeric columns and the label "Unknown" for categorical/text columns. No records were removed during this process.

The comparison verifies whether the number of rows and columns has been maintained and whether missing values have been handled successfully.

```python
# Create the cleaning log to document the issue, affected rows, action, and reason

cleaning_log = pd.DataFrame([
    {
        'Issue': 'Missing numeric values',
        'Affected Rows': int(df[numeric_missing_columns].isnull().sum().sum()),
        'Action Taken': 'Filled missing values using column median',
        'Reason': 'Median is less affected by extreme values than mean'
    },
    {
        'Issue': 'Missing categorical/text values',
        'Affected Rows': int(df[categorical_missing_columns].isnull().sum().sum()),
        'Action Taken': 'Filled missing values with Unknown',
        'Reason': 'Preserves records without guessing category values'
    },
    {
        'Issue': 'Possible extra spaces in text columns',
        'Affected Rows': 'Checked all text columns',
        'Action Taken': 'Removed leading and trailing spaces',
        'Reason': 'Ensures consistent text formatting'
    },
    {
        'Issue': 'Possible Metropolitian spelling variation',
        'Affected Rows': int((df['City'] == 'Metropolitian').sum()),
        'Action Taken': 'Retained original value',
        'Reason': 'Correct spelling needs confirmation from dataset documentation'
    },
    {
        'Issue': 'Negative restaurant coordinates',
        'Affected Rows': int(
            ((df['Restaurant_latitude'] < 0) |
             (df['Restaurant_longitude'] < 0)).sum()
        ),
        'Action Taken': 'Retained values',
        'Reason': 'Negative geographic coordinates can be valid'
    }
])

cleaning_log
```

```python
comparison = pd.DataFrame({
    'Metric': [
        'Total Rows',
        'Total Columns',
        'Missing Values',
        'Duplicate Rows'
    ],
    'Before Cleaning': [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    'After Cleaning': [
        df_clean.shape[0],
        df_clean.shape[1],
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum()
    ]
})

comparison
```
